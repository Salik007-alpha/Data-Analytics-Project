```python
# Data analysis  and Manipulation
import plotly.graph_objs as go
import plotly.io as pio
import plotly.express as px
import pandas as pd

# Data Visulization
import matplotlib.pyplot as plt

#importing plotly
import plotly.offline as py
py.init_notebook_mode(connected=True)

#Initializing plotly
pio.renderers.default = 'colab'
```

```python
salik = pd.read_csv("/content/drive/MyDrive/Data analytics project/Healthcare_Dataset_with_Death___Recovery_Rates.csv")
salik
```

```python
# Return size of the dataframe
salik.size
```

```python
#Return concise summary of data frame
salik.info()
```

```python
# Group by
salik.groupby("Age")["Age"].median()
```

```python
#Return columns labels of the dataset
salik.columns
```

```python
#filtering the Data
filter_country_Region = salik['Age']>50
salik[filter_country_Region]
```

```python
# Drop unwanted columns
salik = salik.drop(['Doctor','Hospital'],axis=1)

#Display the first few row to confirm
salik.sample(5)
```

```python
px.bar(salik.head(15),x = 'Death Rate (%)',
       y = 'Recovery Rate (%)', height = 500)
```

```python
salik.plot(x="Death Rate (%)",y="Recovery Rate (%)",figsize=(18,8))
plt.xlabel("Death Rate (%)")
plt.ylabel("Recovery Rate (%)")
plt.title("Death Rate (%) vs Recovery Rate (%)")
plt.show()
```

```python
salik.plot(x="Death Rate (%)",y="Recovery Rate (%)",color = "black",figsize=(18,8),marker="o",kind = "line")
plt.xlabel("Death Rate (%)")
plt.ylabel("Recovery Rate (%)")
plt.title("Death Rate (%) vs Recovery Rate (%)")
plt.show()
```

```python
salik.plot(x="Death Rate (%)",y="Recovery Rate (%)",color = "black",figsize=(12,5),kind ="bar")
plt.xlabel("Death Rate (%)")
plt.ylabel("Recovery Rate (%)")
plt.title("Death Rate (%) vs Recovery Rate")
plt.show()
```

```python
px.scatter(salik.head(15),x = 'Death Rate (%)',
       y = 'Recovery Rate (%)', color='Death Rate (%)',size_max=60,log_y=True)
```
